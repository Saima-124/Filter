```python
import numpy as np
import matplotlib.pyplot as plt

def six_point_filter(x):
    y = np.zeros(len(x))

    for n in range(len(x)):
        for k in range(6):
            if n - k >= 0:
                y[n] += ((-1) ** k) * x[n-k]

        y[n] = y[n] / 6

    return y
```

```python
# n range
n = np.arange(-10, 51)

# Step input
x1 = np.where(n < 20, 0, 1)

# Filter output
y1 = six_point_filter(x1)

# Plot
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x1)
plt.title("Input 1: Step Signal")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n, y1)
plt.title("Output 1: After 6-Point Differencing Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# n range
n = np.arange(-10, 51)

# Low-frequency input
x2 = np.where(n < 0, 0, np.cos(np.pi * n / 8))

# Filter output
y2 = six_point_filter(x2)

# Plot
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x2)
plt.title("Input 2: Low-Frequency Cosine")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n, y2)
plt.title("Output 2: After 6-Point Differencing Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# n range
n = np.arange(-10, 51)

# High-frequency input
x3 = np.where(n < 0, 0, np.cos(7 * np.pi * n / 8))

# Filter output
y3 = six_point_filter(x3)

# Plot
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x3)
plt.title("Input 3: High-Frequency Cosine")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n, y3)
plt.title("Output 3: After 6-Point Differencing Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

The 6-point differencing filter detects changes or variations in a signal.

It suppresses constant and slowly varying (low-frequency) components.
It gives a stronger response to rapidly changing (high-frequency) components.
Therefore, it behaves as a High-Pass Filter (HPF).

In short:

The 6-point differencing filter acts as a High-Pass Filter that suppresses low-frequency components and emphasizes high-frequency variations.
