```python
import numpy as np
import matplotlib.pyplot as plt

def six_point_averaging_filter(x):
    y = np.zeros(len(x))

    for n in range(len(x)):
        for k in range(6):
            if n - k >= 0:
                y[n] += x[n-k]

        y[n] = y[n] / 6

    return y
```

```python
# n range
n = np.arange(-10, 51)

# Step input
# x(n) = 0, n < 20
#        1, n >= 20

x1 = np.where(n < 20, 0, 1)

# Filter output
y1 = six_point_averaging_filter(x1)

# Plot
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x1)
plt.title("Input 1: Step Signal")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n, y1)
plt.title("Output 1: After 6-Point Averaging Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# n range
n = np.arange(-10, 51)

# Low-frequency input
# x(n) = 0, n < 0
#        cos(pi*n/3), n >= 0

x2 = np.where(n < 0, 0, np.cos(np.pi * n / 8))

# Filter output
y2 = six_point_averaging_filter(x2)

# Plot
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x2)
plt.title("Input 2: Low-Frequency Cosine")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n, y2)
plt.title("Output 2: After 6-Point Averaging Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# n range
n = np.arange(-10, 51)

# High-frequency input
# x(n) = 0, n < 0
#        cos(7*pi*n/3), n >= 0

x3 = np.where(n < 0, 0, np.cos(7 * np.pi * n / 8))

# Filter output
y3 = six_point_averaging_filter(x3)

# Plot
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x3)
plt.title("Input 3: High-Frequency Cosine")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n, y3)
plt.title("Output 3: After 6-Point Averaging Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

The 6-point averaging filter smooths the input signal by taking the average of
six consecutive samples. For the step input, it smooths the sudden transition.
For the cosine inputs, the positive and negative samples cancel during averaging,
so the output is greatly reduced. Therefore, the filter behaves as a smoothing and 
low-pass FIR filter.
