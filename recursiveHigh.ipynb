```python
import numpy as np
import matplotlib.pyplot as plt

# Recursive High-Pass Filter
# y(n) = 0.05y(n-1) + 0.05x(n) - 0.05x(n-1)

def recursive_high_pass(x):
    y = np.zeros(len(x))

    for n in range(len(x)):
        if n == 0:
            y[n] = 0.05 * x[n]
        else:
            y[n] = -0.95 * y[n-1] + 0.05 * x[n] 

    return y
```

```python
n = np.arange(-10, 51)

x1 = np.where(n < 20, 0, 1)

y1 = recursive_high_pass(x1)

plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x1)
plt.title("Input: Step Signal")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid()

plt.subplot(2, 1, 2)
plt.stem(n, y1)
plt.title("Output: Recursive High-Pass Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
n = np.arange(-10, 31)

x2 = np.where(n < 0, 0, np.cos(np.pi * n / 8))

y2 = recursive_high_pass(x2)

plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x2)
plt.title("Input: Low-Frequency Signal")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid()

plt.subplot(2, 1, 2)
plt.stem(n, y2)
plt.title("Output: Recursive High-Pass Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
n = np.arange(-10, 31)

x3 = np.where(n < 0, 0, np.cos(7 * np.pi * n / 8))

y3 = recursive_high_pass(x3)

plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x3)
plt.title("Input: High-Frequency Signal")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid()

plt.subplot(2, 1, 2)
plt.stem(n, y3)
plt.title("Output: Recursive High-Pass Filter")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.grid()

plt.tight_layout()
plt.show()
```

The recursive high-pass filter suppresses slowly varying and low-frequency components while emphasizing rapid changes and high-frequency components. Therefore, it behaves as a High-Pass Filter.
